### **For loop**



```python
for x in "hello":
    print(x)
```

```python
mylist=[100,20,"msd",True,"anand"]
for x in mylist:
    print(x,end=" ")
```

```python
for x in range(10+1): #the 10+1 is called end before
    print(x,end=" ")
```

```python
for x in range(2,10): #this is called start with end before
    print(x,end=" ")
```

```python
for x in range(0,10+1,2): # start,end before,jump
    print(x,end=" ")
```

```python
for x in range(10,2,-2):
    print(x,end=" ")
```

```python
#find a num divisible by 5 in list

mylist=[1,3,6,89,54]
for x in mylist:
    if x%5==0:
        print(x,"divisible by 5")
    else:
        print(x," are not divisible by 5")
```

```python
#nested loop
row=int(input("enter the no of row:"))
col=int(input("enter the nof col:"))
for x in range(row):
    for y in range(col):
        print(y,end=" ")

    print()
    
```

```python
rows=int(input("enter the nof rows:"))
cols=int(input("enter the no of cols:"))
for x in range(rows):
    for y in range(cols):
        print("#",end=" ")
    print()
```

```python
rows=int(input("enter the no of rows:"))
symbol=input("enter the symbol:")

for x in range(rows):
    for y in range(0,x+1):
        print(symbol,end=" ")
    print()
```
