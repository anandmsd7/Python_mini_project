```python
rows= int(input("enter the no of rows:"))
cols=int(input("enter the no of cols:"))
symbols=input("enter the symbol:")
x=0
while x < rows:
    y=0
    while y < cols:
        print(symbols,end=" ")
        y+=1
    print()
    x+=1
```
